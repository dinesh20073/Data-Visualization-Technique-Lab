```python
# ============================================================
# DVT EXPERIMENT – 4
# Name    : B DINESH
# Roll No : 231401025
# Topic   : Healthcare Analytics – Life Expectancy
# ============================================================
import pandas as pd
import numpy as np
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)
df.columns = df.columns.str.strip()
print("Dataset loaded successfully")
print(df.head())
print(df.shape)
print(df.dtypes)
print(df.describe())
# ============================================================
# 1. LIFE EXPECTANCY ABOVE 70
# ============================================================
high_life = df[df["Life expectancy"] > 70]
print(high_life[["Country", "Year", "Life expectancy"]].head(10))
# ============================================================
# 2. GDP ABOVE 10000
# ============================================================
high_gdp = df[df["GDP"] > 10000]
print(high_gdp[["Country", "Year", "GDP", "Life expectancy"]].head(10))
# ============================================================
# 3. HIGHEST LIFE EXPECTANCY
# ============================================================
sorted_df = df.sort_values(
    by="Life expectancy",
    ascending=False
)
print(sorted_df[["Country", "Year", "Life expectancy"]].head(10))
# ============================================================
# 4. DEVELOPED COUNTRIES
# ============================================================
developed = df[df["Status"] == "Developed"]
print(developed[["Country", "Year", "Status", "Life expectancy"]].head(10))
# ============================================================
# 5. LIFE EXPECTANCY STATISTICS
# ============================================================
print("Mean =", df["Life expectancy"].mean())
print("Median =", df["Life expectancy"].median())
print("Mode =", df["Life expectancy"].mode()[0])
print("Standard Deviation =", df["Life expectancy"].std())
# ============================================================
# 6. STATISTICAL SUMMARY
# ============================================================
print(df[["Life expectancy"]].describe())
# ============================================================
# 7. AVERAGE LIFE EXPECTANCY BY STATUS
# ============================================================
status_avg = df.groupby(
    "Status"
)["Life expectancy"].mean()
print(status_avg)
# ============================================================
# 8. AVERAGE LIFE EXPECTANCY BY YEAR
# ============================================================
year_avg = df.groupby(
    "Year"
)["Life expectancy"].mean()
print(year_avg)
# ============================================================
# 9. COUNTRIES WITH LIFE EXPECTANCY BELOW 50
# ============================================================
low_life = df[df["Life expectancy"] < 50]
print(low_life[["Country", "Year", "Life expectancy"]].head(10))
# ============================================================
# 10. HIGHEST SCHOOLING COUNTRIES
# ============================================================
sorted_schooling = df.sort_values(
    by="Schooling",
    ascending=False
)
print(sorted_schooling[["Country", "Year", "Schooling", "Life expectancy"]].head(10))
```
