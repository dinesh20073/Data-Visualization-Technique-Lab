```python
# ============================================================
# DVT EXPERIMENT – 1
# Name    : B DINESH
# Roll No : 231401025
# Topic   : Healthcare Analytics – Life Expectancy
# ============================================================
import pandas as pd
import numpy as np
from google.colab import files
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
# Upload dataset
uploaded = files.upload()
filename = list(uploaded.keys())[0]
df = pd.read_csv(filename)
df.columns = df.columns.str.strip()
print("Dataset Loaded Successfully!")
print(df.head())
print(df.shape)
# ============================================================
# CRISP-DM
# ============================================================
crisp_df = df.copy()
# Business Understanding
# Objective: Predict Life Expectancy
# Data Understanding
print("\n========== CRISP-DM DATA INFORMATION ==========")
print("Dataset Shape:", crisp_df.shape)
print("\nMissing Values:")
print(crisp_df.isnull().sum())
# Data Preparation
crisp_df = crisp_df.dropna(subset=["Life expectancy"])
drop_cols = ["Country"]
for col in drop_cols:
    if col in crisp_df.columns:
        crisp_df.drop(col, axis=1, inplace=True)
le = LabelEncoder()
for col in crisp_df.select_dtypes(include=["object"]).columns:
    crisp_df[col] = le.fit_transform(crisp_df[col].astype(str))
X = crisp_df.drop("Life expectancy", axis=1)
y = crisp_df["Life expectancy"]
imputer = SimpleImputer(strategy="mean")
X = imputer.fit_transform(X)
scaler = StandardScaler()
X = scaler.fit_transform(X)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
# Modeling
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)
# Evaluation
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print("\n========== CRISP-DM RESULTS ==========")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)
print("R2 Score :", r2)
# ============================================================
# SEMMA
# ============================================================
semma_df = df.copy()
# Sample
semma_df = semma_df.dropna(subset=["Life expectancy"])
# Explore
print("\n========== SEMMA DATA INFORMATION ==========")
print(semma_df.info())
# Modify
drop_cols = ["Country"]
for col in drop_cols:
    if col in semma_df.columns:
        semma_df.drop(col, axis=1, inplace=True)
le = LabelEncoder()
for col in semma_df.select_dtypes(include=["object"]).columns:
    semma_df[col] = le.fit_transform(semma_df[col].astype(str))
X = semma_df.drop("Life expectancy", axis=1)
y = semma_df["Life expectancy"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
imputer = SimpleImputer(strategy="mean")
X_train = imputer.fit_transform(X_train)
X_test = imputer.transform(X_test)
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
# Model
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)
# Assess
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print("\n========== SEMMA RESULTS ==========")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)
print("R2 Score :", r2)
# ============================================================
# KDD
# ============================================================
kdd_df = df.copy()
# Step 1: Selection
kdd_df = kdd_df.dropna(subset=["Life expectancy"])
drop_cols = ["Country"]
for col in drop_cols:
    if col in kdd_df.columns:
        kdd_df.drop(col, axis=1, inplace=True)
# Step 2: Preprocessing
le = LabelEncoder()
for col in kdd_df.select_dtypes(include=["object"]).columns:
    kdd_df[col] = le.fit_transform(kdd_df[col].astype(str))
X = kdd_df.drop("Life expectancy", axis=1)
y = kdd_df["Life expectancy"]
imputer = SimpleImputer(strategy="mean")
X = imputer.fit_transform(X)
# Step 3: Transformation
scaler = StandardScaler()
X = scaler.fit_transform(X)
# Step 4: Data Mining
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
model = RandomForestRegressor(random_state=42)
model.fit(X_train, y_train)
# Step 5: Interpretation
y_pred = model.predict(X_test)
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, y_pred)
print("\n========== KDD RESULTS ==========")
print("MAE :", mae)
print("MSE :", mse)
print("RMSE :", rmse)
print("R2 Score :", r2)
```
