```python
# ============================================================
# DVT EXPERIMENT – 2
# Name    : B DINESH
# Roll No : 231401025
# Topic   : Healthcare Analytics – Life Expectancy
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)
df.columns = df.columns.str.strip()
# ============================================================
# 1. BASIC DATASET INFORMATION
# ============================================================
print("========== DATASET INFORMATION ==========")
print("\nDataset Shape:")
print(df.shape)
print("\nFirst 5 Rows:")
display(df.head())
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
print("\nStatistical Summary:")
display(df.describe())
# ============================================================
# 2. TARGET VARIABLE DISTRIBUTION
# ============================================================
plt.figure(figsize=(7, 5))
sns.histplot(x=df["Life expectancy"], data=df, bins=20, kde=True)
plt.title("Life Expectancy Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Life Expectancy")
plt.ylabel("Number of Countries")
plt.show()
# ============================================================
# 3. ADULT MORTALITY DISTRIBUTION
# ============================================================
plt.figure(figsize=(8, 5))
sns.histplot(df["Adult Mortality"], bins=20, kde=True)
plt.title("Adult Mortality Distribution", fontsize=16, fontweight="bold")
plt.xlabel("Adult Mortality")
plt.ylabel("Number of Countries")
plt.show()
# ============================================================
# 4. GDP DISTRIBUTION
# ============================================================
plt.figure(figsize=(8, 5))
sns.histplot(df["GDP"], bins=25, kde=True)
plt.title("GDP Distribution", fontsize=16, fontweight="bold")
plt.xlabel("GDP")
plt.ylabel("Frequency")
plt.show()
# ============================================================
# 5. STATUS VS LIFE EXPECTANCY
# ============================================================
plt.figure(figsize=(8, 5))
sns.boxplot(x="Status", y="Life expectancy", data=df)
plt.title("Development Status vs Life Expectancy", fontsize=16, fontweight="bold")
plt.xlabel("Status")
plt.ylabel("Life Expectancy")
plt.show()
# ============================================================
# 6. SCHOOLING VS LIFE EXPECTANCY
# ============================================================
plt.figure(figsize=(8, 5))
sns.scatterplot(x="Schooling", y="Life expectancy", hue="Status", data=df)
plt.title("Schooling vs Life Expectancy", fontsize=16, fontweight="bold")
plt.xlabel("Schooling")
plt.ylabel("Life Expectancy")
plt.legend(title="Status")
plt.show()
# ============================================================
# 7. GDP VS DEVELOPMENT STATUS
# ============================================================
plt.figure(figsize=(8, 5))
sns.boxplot(x="Status", y="GDP", data=df)
plt.title("GDP vs Development Status", fontsize=16, fontweight="bold")
plt.xlabel("Status")
plt.ylabel("GDP")
plt.show()
# ============================================================
# 8. CORRELATION HEATMAP
# ============================================================
plt.figure(figsize=(12, 8))
correlation = df.select_dtypes(include="number").corr()
sns.heatmap(correlation, annot=True, fmt=".2f", cmap="coolwarm", linewidths=0.5)
plt.title("Correlation Heatmap of Life Expectancy Dataset", fontsize=18, fontweight="bold")
plt.show()
# ============================================================
# 9. FEATURE-WISE BOXPLOTS
# ============================================================
features = ["Life expectancy", "Adult Mortality", "Alcohol", "BMI", "GDP"]
for feature in features:
    plt.figure(figsize=(7, 5))
    sns.boxplot(y=df[feature])
    plt.title(f"Boxplot of {feature}", fontsize=16, fontweight="bold")
    plt.ylabel(feature)
    plt.show()
# ============================================================
# 10. FINAL SUMMARY
# ============================================================
print("\n========== EXPERIMENT 2 SUMMARY ==========")
print("Dataset Shape:", df.shape)
print("Number of Features:", len(df.columns) - 1)
print("Target Column: Life expectancy")
print("Total Countries:", len(df))
print("Average Life Expectancy:", df["Life expectancy"].mean())
print("Maximum Life Expectancy:", df["Life expectancy"].max())
print("Minimum Life Expectancy:", df["Life expectancy"].min())
print("\nExperiment 2 Completed Successfully!")
```
