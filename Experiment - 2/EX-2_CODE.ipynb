```python
# ============================================================
# DVT EXPERIMENT – 2
# Name    : B DINESH
# Roll No : 231401025
# Topic   : Healthcare Analytics – Life Expectancy
# ============================================================
import pandas as pd
import matplotlib.pyplot as plt
from google.colab import files
uploaded = files.upload()
file_name = next(iter(uploaded))
df = pd.read_csv(file_name)
df.columns = df.columns.str.strip()
df.head()
# ---------------------------------------------------
# 1. TOP 10 COUNTRIES BY AVERAGE LIFE EXPECTANCY
# ---------------------------------------------------
top_countries_life = (
    df.groupby("Country")["Life expectancy"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
print("Top 10 Countries by Average Life Expectancy")
display(top_countries_life)
plt.figure(figsize=(10,5))
plt.bar(top_countries_life["Country"], top_countries_life["Life expectancy"])
plt.title("Top 10 Countries by Average Life Expectancy")
plt.xlabel("Country")
plt.ylabel("Average Life Expectancy")
plt.xticks(rotation=45)
plt.show()
# ---------------------------------------------------
# 2. TOP 10 COUNTRIES BY AVERAGE GDP
# ---------------------------------------------------
top_countries_gdp = (
    df.groupby("Country")["GDP"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
print("Top 10 Countries by Average GDP")
display(top_countries_gdp)
plt.figure(figsize=(10,5))
plt.bar(top_countries_gdp["Country"], top_countries_gdp["GDP"])
plt.title("Top 10 Countries by Average GDP")
plt.xlabel("Country")
plt.ylabel("Average GDP")
plt.xticks(rotation=45)
plt.show()
# ---------------------------------------------------
# 3. TOP 10 COUNTRIES BY AVERAGE SCHOOLING
# ---------------------------------------------------
top_countries_schooling = (
    df.groupby("Country")["Schooling"]
    .mean()
    .sort_values(ascending=False)
    .head(10)
    .reset_index()
)
print("Top 10 Countries by Average Schooling")
display(top_countries_schooling)
plt.figure(figsize=(12,5))
plt.bar(top_countries_schooling["Country"], top_countries_schooling["Schooling"])
plt.title("Top 10 Countries by Average Schooling")
plt.xlabel("Country")
plt.ylabel("Average Schooling")
plt.xticks(rotation=45)
plt.show()
# ---------------------------------------------------
# 4. DEVELOPMENT STATUS DISTRIBUTION
# ---------------------------------------------------
status_type = (
    df["Status"]
    .value_counts()
    .reset_index()
)
status_type.columns = ["Status", "Count"]
print("Development Status Distribution")
display(status_type)
plt.figure(figsize=(8,5))
plt.bar(status_type["Status"], status_type["Count"])
plt.title("Development Status Distribution")
plt.xlabel("Status")
plt.ylabel("Number of Records")
plt.show()
```
