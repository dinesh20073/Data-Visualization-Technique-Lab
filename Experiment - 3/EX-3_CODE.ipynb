```python
# ============================================================
# DVT EXPERIMENT – 3
# Name    : B DINESH
# Roll No : 231401025
# Topic   : Healthcare Analytics – Life Expectancy
# ============================================================
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler
df = pd.read_csv("Life Expectancy Data.csv")
df.columns = df.columns.str.strip()
print("========== DATASET INFORMATION ==========")
print("\nDataset Shape:")
print(df.shape)
print("\nFirst 5 Rows:")
display(df.head())
print("\nDataset Information:")
df.info()
print("\nMissing Values:")
print(df.isnull().sum())
print("\nStatistical Summary:")
display(df.describe())
# ============================================================
# 1. HANDLE MISSING VALUES
# ============================================================
df["Life expectancy"] = df["Life expectancy"].fillna(df["Life expectancy"].mean())
df["Adult Mortality"] = df["Adult Mortality"].fillna(df["Adult Mortality"].mean())
df["Alcohol"] = df["Alcohol"].fillna(df["Alcohol"].mean())
df["Hepatitis B"] = df["Hepatitis B"].fillna(df["Hepatitis B"].mean())
df["BMI"] = df["BMI"].fillna(df["BMI"].mean())
df["Polio"] = df["Polio"].fillna(df["Polio"].mean())
df["Total expenditure"] = df["Total expenditure"].fillna(df["Total expenditure"].mean())
df["Diphtheria"] = df["Diphtheria"].fillna(df["Diphtheria"].mean())
df["GDP"] = df["GDP"].fillna(df["GDP"].mean())
df["Population"] = df["Population"].fillna(df["Population"].mean())
df["thinness  1-19 years"] = df["thinness  1-19 years"].fillna(df["thinness  1-19 years"].mean())
df["thinness 5-9 years"] = df["thinness 5-9 years"].fillna(df["thinness 5-9 years"].mean())
df["Income composition of resources"] = df["Income composition of resources"].fillna(df["Income composition of resources"].mean())
df["Schooling"] = df["Schooling"].fillna(df["Schooling"].mean())
print("\nMissing Values After Imputation:")
print(df.isnull().sum())
# ============================================================
# 2. REMOVE DUPLICATE RECORDS
# ============================================================
print("\nDuplicate Records:", df.duplicated().sum())
df.drop_duplicates(inplace=True)
print("Duplicate Records After Removal:", df.duplicated().sum())
# ============================================================
# 3. OUTLIER DETECTION
# ============================================================
Q1 = df["Life expectancy"].quantile(0.25)
Q3 = df["Life expectancy"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["Life expectancy"] < lower) | (df["Life expectancy"] > upper)]
print("\n========== OUTLIERS ==========")
print(outliers[["Country", "Year", "Life expectancy"]])
# ============================================================
# 4. DATA NORMALIZATION
# ============================================================
scaler = MinMaxScaler()
df["Life_expectancy_Normalized"] = scaler.fit_transform(df[["Life expectancy"]])
print("\n========== NORMALIZED DATA ==========")
display(df[["Life expectancy", "Life_expectancy_Normalized"]].head())
# ============================================================
# 5. FINAL SUMMARY
# ============================================================
print("\n========== EXPERIMENT 3 SUMMARY ==========")
print("Dataset Shape:", df.shape)
print("Total Records:", len(df))
print("Duplicate Records:", df.duplicated().sum())
print("Number of Life Expectancy Outliers:", len(outliers))
print("Life Expectancy Normalization Completed")
print("\nExperiment 3 Completed Successfully!")
```
