```python
# ============================================================
# IMPORT LIBRARIES AND LOAD DATASET
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from google.colab import files

# Upload dataset
uploaded = files.upload()

# Read dataset
df = pd.read_csv("Life Expectancy Data.csv")

# Remove extra spaces from column names
df.columns = df.columns.str.strip()

print("Dataset loaded successfully!")


# ============================================================
# DATA INSPECTION
# ============================================================

# Display first five records
print("\nFirst 5 Records:")
display(df.head())

# Display last five records
print("\nLast 5 Records:")
display(df.tail())

# Display dataset shape
print("\nDataset Shape:")
print(df.shape)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)

# Display dataset information
print("\nDataset Information:")
df.info()

# Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe())

# Check missing values
print("\nMissing Values:")
print(df.isnull().sum())


# ============================================================
# DATA CLEANING
# ============================================================

# Fill missing numerical values with the median
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())

# Fill missing categorical values with the mode
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    if df[column].isnull().sum() > 0:
        df[column] = df[column].fillna(df[column].mode()[0])

print("\nMissing values after cleaning:")
print(df.isnull().sum())


# ============================================================
# FILTERING DATA
# ============================================================

# Select records where Life Expectancy is greater than 70
high_life_expectancy = df[df["Life expectancy"] > 70]

print("\nRecords with Life Expectancy Greater Than 70:")
display(
    high_life_expectancy[
        ["Country", "Year", "Status", "Life expectancy"]
    ].head(10)
)

# Select records where GDP is greater than 10000
high_gdp = df[df["GDP"] > 10000]

print("\nRecords with GDP Greater Than 10000:")
display(
    high_gdp[
        ["Country", "Year", "GDP", "Life expectancy"]
    ].head(10)
)


# ============================================================
# SORTING DATA
# ============================================================

# Sort Life Expectancy in descending order
sorted_df = df.sort_values(
    by="Life expectancy",
    ascending=False
)

print("\nLife Expectancy Sorted in Descending Order:")
display(
    sorted_df[
        ["Country", "Year", "Life expectancy"]
    ].head(10)
)


# ============================================================
# VISUALIZATION – LIFE EXPECTANCY RANKING
# ============================================================

# Select top 10 records
top_10 = sorted_df.head(10)

plt.figure(figsize=(12, 6))

plt.bar(
    top_10["Country"],
    top_10["Life expectancy"]
)

plt.xlabel("Country")
plt.ylabel("Life Expectancy")
plt.title("Life Expectancy Ranking")

plt.xticks(rotation=45)
plt.tight_layout()
plt.show()


# ============================================================
# STATISTICAL ANALYSIS
# ============================================================

# Calculate mean
mean_value = df["Life expectancy"].mean()

# Calculate median
median_value = df["Life expectancy"].median()

# Calculate mode
mode_value = df["Life expectancy"].mode()

# Calculate standard deviation
std_value = df["Life expectancy"].std()

print("\nStatistical Analysis of Life Expectancy:")

print("Mean =", round(mean_value, 2))
print("Median =", round(median_value, 2))
print("Mode =", mode_value.tolist())
print("Standard Deviation =", round(std_value, 2))

# Mean Formula:
# Mean = Sum of all values / Number of values


# ============================================================
# DESCRIPTIVE STATISTICS
# ============================================================

print("\nDescriptive Statistics of Life Expectancy:")

print(df["Life expectancy"].describe())


# ============================================================
# GROUPING AND AGGREGATION
# ============================================================

# Calculate average Life Expectancy by Status
status_avg = df.groupby(
    "Status"
)["Life expectancy"].mean()

print("\nAverage Life Expectancy by Country Status:")
print(status_avg)


# ============================================================
# VISUALIZATION – STATUS-WISE AVERAGE LIFE EXPECTANCY
# ============================================================

plt.figure(figsize=(8, 5))

status_avg.plot(kind="bar")

plt.xlabel("Country Status")
plt.ylabel("Average Life Expectancy")
plt.title("Status-wise Average Life Expectancy")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()


# ============================================================
# COUNTRY-WISE AVERAGE LIFE EXPECTANCY
# ============================================================

# Calculate average Life Expectancy for each country
country_avg = (
    df.groupby("Country")["Life expectancy"]
    .mean()
    .sort_values(ascending=False)
)

print("\nTop 10 Countries by Average Life Expectancy:")
print(country_avg.head(10))


# ============================================================
# CORRELATION ANALYSIS
# ============================================================

# Select numerical columns
numeric_df = df.select_dtypes(include=np.number)

# Calculate correlation matrix
corr = numeric_df.corr()

print("\nCorrelation Matrix:")
display(corr)


# ============================================================
# CORRELATION WITH LIFE EXPECTANCY
# ============================================================

# Display correlation values with Life Expectancy
life_corr = corr[
    "Life expectancy"
].sort_values(ascending=False)

print("\nCorrelation with Life Expectancy:")
print(life_corr)


# ============================================================
# CORRELATION HEATMAP
# ============================================================

# Display correlation heatmap
plt.figure(figsize=(14, 10))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap – Life Expectancy Dataset")
plt.tight_layout()
plt.show()


# ============================================================
# SCHOOLING VS LIFE EXPECTANCY
# ============================================================

# Visualize the relationship between Schooling and Life Expectancy
plt.figure(figsize=(8, 6))

plt.scatter(
    df["Schooling"],
    df["Life expectancy"]
)

plt.xlabel("Schooling")
plt.ylabel("Life Expectancy")
plt.title("Schooling vs Life Expectancy")

plt.tight_layout()
plt.show()


# ============================================================
# GDP VS LIFE EXPECTANCY
# ============================================================

# Visualize the relationship between GDP and Life Expectancy
plt.figure(figsize=(8, 6))

plt.scatter(
    df["GDP"],
    df["Life expectancy"]
)

plt.xlabel("GDP")
plt.ylabel("Life Expectancy")
plt.title("GDP vs Life Expectancy")

plt.tight_layout()
plt.show()


# ============================================================
# EXPERIMENT 4B SUMMARY
# ============================================================

print("\n" + "=" * 60)
print("DVT EXPERIMENT 4B SUMMARY")
print("=" * 60)

print("Dataset Shape:", df.shape)
print("Total Records:", len(df))
print("Total Features:", len(df.columns))

print(
    "\nMean Life Expectancy:",
    round(df["Life expectancy"].mean(), 2)
)

print(
    "Median Life Expectancy:",
    round(df["Life expectancy"].median(), 2)
)

print(
    "Standard Deviation:",
    round(df["Life expectancy"].std(), 2)
)

print("\nExperiment 4B Completed Successfully!")
print("=" * 60)
```
